En premier lieu un jeu que l’on notera A qui est un simple jeu de pile ou face avec une pièce biaisée (pile avec une probabilité de p=0.49). Le joueur mise un euro et lance la pièce : s’il obtient pile il gagne un euro en plus de récupérer sa mise, sinon il perd sa mise.
Ensuite, un jeu que l’on notera B, qui est un jeu avec deux pièces biaisées. La première pièce donne pile avec une probabilité p1 = 0.09 et la seconde pièce donne pile avec une probabilité p2 = 0.74. Le joueur ne peut miser qu’un euro à la fois ! En revanche, on regarde à chaque lancé son capital (la somme d’argent totale) dont il dispose pour déterminer quelle pièce lancer : si le capital est un multiple de 3, on lance la pièce numéro une, sinon on lance la seconde pièce. Comme dans le jeu A, le joueur remporte sa mise plus un euro supplémentaire si la pièce choisie tombe sur pile, sinon il perd sa mise.

```python


resultatsJeu1 = []
resultatsJeu2 = []

for j in range(10000):
    capital = 1000
    for i in range(200):
        if random.binomialvariate(1, 0.49)==1:
            capital+=1
        else:
            capital-=1

    resultatsJeu1.append(capital)

    capital = 1000
    for i in range(200):
        if (i%3 == 0):
            if random.binomialvariate(1, 0.09)==1:
                capital+=1
            else:
                capital-=1
        else:
            if random.binomialvariate(1, 0.74)==1:
                capital+=1
            else:
                capital-=1
    resultatsJeu2.append(capital)

print("Jeu1: "+ str(sum(resultatsJeu1)/len(resultatsJeu1)))
print("Jeu2: "+ str(sum(resultatsJeu2)/len(resultatsJeu2)))
```
